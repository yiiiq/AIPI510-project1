```python

import pandas as pd

# Load full dataset
df = pd.read_csv("../data/SVI_2022_US_county.csv")

# Define Gulf Coast states
gulf_states = ["TX", "LA", "MS", "AL", "FL"]

# Filter only Gulf Coast counties
df_gulf = df[df["ST_ABBR"].isin(gulf_states)].copy()

# Columns to keep: ID, overall, themes, key hurricane-related flags
columns_to_keep = [
    # Identification
     "STATE", "COUNTY", "FIPS",
    
    # Overall and theme scores
    "RPL_THEMES", "RPL_THEME1", "RPL_THEME2", "RPL_THEME3", "RPL_THEME4",
    
    # Key vulnerability flags for hurricanes
    "F_NOVEH",     # households with no vehicles
    "F_MOBILE",    # mobile homes
    "F_CROWD",     # household crowding
    "F_UNINSUR",   # uninsured
    "F_AGE65",     # elderly
    "F_DISABL",    # disabled
    "F_SNGPNT"     # single-parent households
]

df_gulf_hurricane = df_gulf[columns_to_keep]

# Quick check
print("Shape:", df_gulf_hurricane.shape)
df_gulf_hurricane.head(20)
```
